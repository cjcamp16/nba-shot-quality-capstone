# Feature Relationships

This notebook explores relationships between shot-level features and shooting success.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Loading in Shots datasets

seasons = [
    "2013-14","2014-15","2015-16","2016-17",
    "2017-18","2018-19","2019-20","2020-21",
    "2021-22","2022-23","2023-24","2024-25",
    "2025-26"
]

dfs = []

for season in seasons:
    temp = pd.read_parquet(
        f"data/raw/shots/shots_{season}.parquet"
    )
    dfs.append(temp)

shots = pd.concat(dfs, ignore_index=True)
```

```python
# Loading Games Datasets

game_dfs = []

for season in seasons:
    temp = pd.read_parquet(
        f"data/raw/games/games_{season}.parquet"
    )
    game_dfs.append(temp)

games = pd.concat(game_dfs, ignore_index=True)
```

# Winning vs Losing Team Shot Profiles and FG% averaged from 2013-2026

```python
shots_games = shots.merge(
    games[["GAME_ID", "TEAM_ID", "WL"]],
    on=["GAME_ID", "TEAM_ID"],
    how="left"
)
```

```python
attempt_profile = pd.crosstab(
    shots_games["WL"],
    shots_games["SHOT_ZONE_BASIC"],
    normalize="index"
) * 100

fg_profile = (
    shots_games.groupby(
        ["WL", "SHOT_ZONE_BASIC"]
    )["SHOT_MADE_FLAG"]
    .mean()
    .mul(100)
    .unstack()
)
```

```python
comparison_table = pd.DataFrame({
    "W Attempt %": attempt_profile.loc["W"],
    "L Attempt %": attempt_profile.loc["L"],
    "Attempt % Diff": (
        attempt_profile.loc["W"]
        - attempt_profile.loc["L"]
    ),

    "W FG%": fg_profile.loc["W"],
    "L FG%": fg_profile.loc["L"],
    "FG% Diff": (
        fg_profile.loc["W"]
        - fg_profile.loc["L"]
    )
})

comparison_table = comparison_table.round(2)

comparison_table
```

# Winning vs Losing Team Shot Profiles and FG% from 2013 compared to 2026 

```python
shots_2013 = pd.read_parquet("data/raw/shots/shots_2013-14.parquet")
games_2013 = pd.read_parquet("data/raw/games/games_2013-14.parquet")

shots_2026 = pd.read_parquet("data/raw/shots/shots_2025-26.parquet")
games_2026 = pd.read_parquet("data/raw/games/games_2025-26.parquet")
```

```python
shots_games_2013 = shots_2013.merge(games_2013[["GAME_ID", "TEAM_ID", "WL"]], on=["GAME_ID", "TEAM_ID"], how="left")

shots_games_2026 = shots_2026.merge(games_2026[["GAME_ID", "TEAM_ID", "WL"]], on=["GAME_ID", "TEAM_ID"], how="left")
```

```python
attempt_profile_2013 = pd.crosstab(shots_games_2013["WL"],shots_games_2013["SHOT_ZONE_BASIC"],normalize="index") * 100

fg_profile_2013 = (shots_games_2013.groupby(["WL", "SHOT_ZONE_BASIC"])["SHOT_MADE_FLAG"].mean().mul(100).unstack())

comparison_table_2013 = pd.DataFrame({
    "W Attempt %": attempt_profile_2013.loc["W"],
    "L Attempt %": attempt_profile_2013.loc["L"],
    "Attempt % Diff": (
        attempt_profile_2013.loc["W"]
        - attempt_profile_2013.loc["L"]
    ),
    "W FG%": fg_profile_2013.loc["W"],
    "L FG%": fg_profile_2013.loc["L"],
    "FG% Diff": (
        fg_profile_2013.loc["W"]
        - fg_profile_2013.loc["L"]
    )
}).round(2)
```

```python
attempt_profile_2026 = pd.crosstab(shots_games_2026["WL"], shots_games_2026["SHOT_ZONE_BASIC"], normalize="index") * 100

fg_profile_2026 = (shots_games_2026.groupby(["WL", "SHOT_ZONE_BASIC"])["SHOT_MADE_FLAG"].mean().mul(100).unstack())

comparison_table_2026 = pd.DataFrame({
    "W Attempt %": attempt_profile_2026.loc["W"],
    "L Attempt %": attempt_profile_2026.loc["L"],
    "Attempt % Diff": (
        attempt_profile_2026.loc["W"]
        - attempt_profile_2026.loc["L"]
    ),
    "W FG%": fg_profile_2026.loc["W"],
    "L FG%": fg_profile_2026.loc["L"],
    "FG% Diff": (
        fg_profile_2026.loc["W"]
        - fg_profile_2026.loc["L"]
    )
}).round(2)
```
