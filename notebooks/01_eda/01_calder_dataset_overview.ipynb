# Dataset Overview

Initial exploratory analysis of our NBA Datasets.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# Individual Games Dataset From Each Season (2013-2026)

```python
# Example
df = pd.read_parquet("data/raw/games/games_2019-20.parquet")
df.head()
```

```python
# Graph Depicting Variable Type in the Games Dataset

game_categories = {
    "Scoring": ["PTS"],
    "Shooting Volume": ["FGA", "FG3A", "FTA"],
    "Made Shots": ["FGM", "FG3M", "FTM"],
    "Rebounding": ["OREB", "DREB", "REB"],
    "Playmaking": ["AST"],
    "Defense": ["STL", "BLK"],
    "Mistakes/Fouls": ["TOV", "PF"]
}

category_counts = {category: len(columns) for category, columns in game_categories.items()}

pd.Series(category_counts).plot(kind="bar", figsize=(10, 5), color="navy")

plt.title("Types of Variables in the Games Dataset")
plt.xlabel("Variable Category")
plt.ylabel("Number of Variables")
plt.xticks(rotation=45)
plt.show()
```

# Individual Shots Dataset From Each Season (2013-2026)

```python
# Example
df1 = pd.read_parquet("data/raw/shots/shots_2013-14.parquet")
df1.head()
```

```python
shot_categories = {
    "Location": 4,          
    "Shot Type": 3,         
    "Timing": 3,            
    "Defense": 2,          
    "Game Context": 3,     
    "Player Context": 2,    
    "Outcome": 1            
}

pd.Series(shot_categories).sort_values(ascending=False).plot(kind="bar", figsize=(10,5), color="navy")

plt.title("Types of Variables in the Shot Dataset")
plt.xlabel("Variable Category")
plt.ylabel("Number of Variables")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# Players Dataset

```python
# Example
df2 = pd.read_parquet("data/raw/players.parquet")
df2.head()
```

```python
player_categories = {
    "Personal Information": 4,    # first name, last name, birthdate 
    "Physical Attributes": 2,     # height, weight
    "Career Information": 3,      # from year, to year, seasons
    "Roster Information": 3,      # team, position, status
    "Identifiers": 2              # player_id, team_id
}

pd.Series(player_categories).sort_values(ascending=False).plot(
    kind="bar",
    figsize=(10,5), color ="navy"
)

plt.title("Types of Variables in the Players Dataset")
plt.xlabel("Variable Category")
plt.ylabel("Number of Variables")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# Teams dataset

```python
# Example
df3 = pd.read_parquet("data/raw/teams.parquet")
df3.head()
```

```python
team_categories = {
    "Team Identity": 4,        # team_id, team_name, abbreviation, nickname 
    "Geographic Information": 2, # city, state
    "League Structure": 2,     # conference, division
    "Historical Information": 2 # year founded
}

pd.Series(team_categories).sort_values(ascending=False).plot(kind="bar",figsize=(10,5), color = "navy")

plt.title("Types of Variables in the Teams Dataset")
plt.xlabel("Variable Category")
plt.ylabel("Number of Variables")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
